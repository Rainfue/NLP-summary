### 1. Bag of Words

```python
x1 = np.array([1,1,0,1,1,3,0,3,1,3])
x1 = np.array(x1/sum(x1)).reshape(1,-1)
print(x1)

x2 = np.array([1,0,1,0,0,3,1,3,0,2])
x2 = np.array(x2/sum(x2)).reshape(1,-1)
print(x2)
print('--')
print(x2.sum())
```

```python
# выводим схожесть двух кусочков рикролла
cosine_similarity(x1, x2)
```
